```python
all_quakes = []

start = datetime.datetime(2010, 1, 1)
end = datetime.datetime(2026, 1, 1)
delta = datetime.timedelta(days=365)  # one year at a time

current = start
while current < end:
    next_date = min(current + delta, end)
    params = {
        'format': 'geojson',
        'starttime': current.strftime('%Y-%m-%d'),
        'endtime': next_date.strftime('%Y-%m-%d'),
        'minmagnitude': 4.5,
        'orderby': 'time'
    }
    res = requests.get('https://earthquake.usgs.gov/fdsnws/event/1/query', params=params).json()
    all_quakes.extend(res['features'])
    current = next_date

quakes = pd.DataFrame([q['properties'] for q in all_quakes])
```

# Final Analysis — Solar Flare Activity vs. Terrestrial Phenomena

```python
from scipy.stats import pearsonr, spearmanr
from statsmodels.stats.multitest import multipletests

START = pd.Timestamp("2016-01-01")
END   = pd.Timestamp("2025-12-31")

def daily_date(values):
    return (
        pd.to_datetime(values, errors="coerce", utc=True)
        .dt.tz_localize(None)
        .dt.normalize()
    )

cal = pd.DataFrame({
    "date": pd.date_range(START, END, freq="D")
})
```

```python
sf = sol.copy()

sf["date"] = daily_date(sf["peakTime"])

sf = sf[sf["date"].between(START, END)].copy()

sf["letter"] = sf["classType"].astype(str).str[0].str.upper()

sf["number"] = pd.to_numeric(
    sf["classType"].astype(str).str[1:],
    errors="coerce"
)

class_flux = {
    "A": 1e-8,
    "B": 1e-7,
    "C": 1e-6,
    "M": 1e-5,
    "X": 1e-4
}

sf["flux"] = (
    sf["letter"].map(class_flux) *
    sf["number"]
)

solar_daily_final = (
    sf.groupby("date")
      .agg(
          flare_count=("flrID", "count"),
          total_flux=("flux", "sum"),
          max_flux=("flux", "max"),
          x_count=("letter", lambda x: (x == "X").sum()),
          m_count=("letter", lambda x: (x == "M").sum())
      )
      .reset_index()
)

solar_final = cal.merge(
    solar_daily_final,
    on="date",
    how="left"
).fillna(0)
```

```python
eq = quakes.copy()

eq["date"] = daily_date(eq["time"])

eq = eq[eq["date"].between(START, END)].copy()

eq_daily_final = (
    eq.groupby("date")
      .agg(
          earthquake_count=("mag", "size"),
          large_earthquake_count=(
              "mag",
              lambda x: (x >= 6.0).sum()
          )
      )
      .reset_index()
)

eq_final = cal.merge(
    eq_daily_final,
    on="date",
    how="left"
).fillna(0)
```

```python
solar_eq_final = solar_final.merge(
    eq_final,
    on="date",
    how="left"
).fillna(0)
```

```python
def correlation_by_lag(
    df,
    solar_column,
    terrestrial_column,
    max_lag=7
):
    results = []

    for lag in range(max_lag + 1):

        x = df[solar_column]
        y = df[terrestrial_column].shift(-lag)

        valid = x.notna() & y.notna()

        x_valid = x[valid]
        y_valid = y[valid]

        if len(x_valid) < 3:
            continue

        pearson_r, pearson_p = pearsonr(
            x_valid,
            y_valid
        )

        spearman_r, spearman_p = spearmanr(
            x_valid,
            y_valid
        )

        results.append({
            "lag_days": lag,
            "pearson_r": pearson_r,
            "pearson_p": pearson_p,
            "spearman_r": spearman_r,
            "spearman_p": spearman_p
        })

    result = pd.DataFrame(results)

    if len(result) > 0:

        result["pearson_fdr"] = multipletests(
            result["pearson_p"],
            method="fdr_bh"
        )[1]

        result["spearman_fdr"] = multipletests(
            result["spearman_p"],
            method="fdr_bh"
        )[1]

    return result
```

```python
earthquake_results = correlation_by_lag(
    solar_eq_final,
    "flare_count",
    "earthquake_count",
    max_lag=7
)

large_earthquake_results = correlation_by_lag(
    solar_eq_final,
    "flare_count",
    "large_earthquake_count",
    max_lag=7
)

print("Solar flare count → all earthquakes")
display(earthquake_results)

print("\nSolar flare count → M6+ earthquakes")
display(large_earthquake_results)
```

```python
flux_results = correlation_by_lag(
    solar_eq_final,
    "total_flux",
    "earthquake_count",
    max_lag=7
)

print("\nSolar flare energy proxy → all earthquakes")
display(flux_results)
```

```python
xflare_results = correlation_by_lag(
    solar_eq_final,
    "x_count",
    "earthquake_count",
    max_lag=7
)

print("\nX-class flare count → all earthquakes")
display(xflare_results)
```

```python
def daily_date(values):
    """
    Convert datetime-like values to timezone-naive,
    normalized daily timestamps.
    """
    return (
        pd.to_datetime(values, errors="coerce", utc=True)
        .dt.tz_localize(None)
        .dt.normalize()
    )


def correlation_by_lag(
    df,
    solar_column,
    earth_column,
    max_lag=7
):
    """
    Calculate Pearson and Spearman correlations
    from lag 0 through max_lag.

    Positive lag means precipitation occurs AFTER
    the solar activity.
    """

    results = []

    for lag in range(max_lag + 1):

        x = df[solar_column]
        y = df[earth_column].shift(-lag)

        valid = (
            x.notna() &
            y.notna() &
            np.isfinite(x) &
            np.isfinite(y)
        )

        x_valid = x[valid]
        y_valid = y[valid]

        if len(x_valid) < 3:
            continue

        if x_valid.nunique() < 2 or y_valid.nunique() < 2:
            continue

        pearson_r, pearson_p = pearsonr(
            x_valid,
            y_valid
        )

        spearman_r, spearman_p = spearmanr(
            x_valid,
            y_valid
        )

        results.append({
            "lag_days": lag,
            "n": len(x_valid),

            "pearson_r": pearson_r,
            "pearson_p": pearson_p,

            "spearman_r": spearman_r,
            "spearman_p": spearman_p
        })

    result = pd.DataFrame(results)

    if len(result) > 0:

        result["pearson_fdr"] = multipletests(
            result["pearson_p"],
            method="fdr_bh"
        )[1]

        result["spearman_fdr"] = multipletests(
            result["spearman_p"],
            method="fdr_bh"
        )[1]

    return result
```

```python
possible_temp_cols = [
    "temp_anomaly",
    "temperature_anomaly",
    "Temperature Anomaly",
    "temperature",
    "Temperature",
    "temp",
    "Temp",
    "anomaly",
    "Anomaly"
]


temp_col = None


for col in possible_temp_cols:

    if col in temp.columns:

        temp_col = col
        break

if temp_col is None:

    for col in temp.columns:

        lower_col = str(col).lower()

        if (
            "anom" in lower_col
            or "temp" in lower_col
        ):

            if pd.api.types.is_numeric_dtype(
                temp[col]
            ):

                temp_col = col
                break


if temp_col is None:

    numeric_cols = (
        temp
        .select_dtypes(
            include=np.number
        )
        .columns
        .tolist()
    )

    print("\nNumeric columns found:")
    print(numeric_cols)

    raise ValueError(
        "Could not automatically identify "
        "the temperature anomaly column."
    )


print(
    f"Using temperature column: "
    f"{temp_col}"
)


temp["temp_anomaly"] = pd.to_numeric(
    temp[temp_col],
    errors="coerce"
)
```

```python
temp = temp[
    temp["date"].between(
        START,
        END
    )
].copy()


temp = temp.dropna(
    subset=[
        "date",
        "temp_anomaly"
    ]
)

temp_daily_final = (
    temp
    .groupby(
        "date",
        as_index=False
    )
    ["temp_anomaly"]
    .mean()
)


print("\nPrepared temperature sample:")

display(
    temp_daily_final.head()
)


print(
    "\nNumber of temperature days:",
    len(temp_daily_final)
)


print(
    "Temperature date range:",
    temp_daily_final["date"].min(),
    "to",
    temp_daily_final["date"].max()
)
```

```python
solar_temp = solar_final.merge(
    temp_daily_final,
    on="date",
    how="inner"
)


print("\n" + "=" * 70)
print("MERGED DATA")
print("=" * 70)


print(
    "\nDays available for analysis:",
    len(solar_temp)
)
```

```python
temp_flare_results = correlation_by_lag(
    solar_temp,
    "flare_count",
    "temp_anomaly",
    max_lag=7
)


print("\n")
print("=" * 70)

print(
    "SOLAR FLARE COUNT → "
    "GLOBAL TEMPERATURE ANOMALY"
)

print("=" * 70)

display(
    temp_flare_results
)
```

```python
temp_energy_results = correlation_by_lag(
    solar_temp,
    "total_flux",
    "temp_anomaly",
    max_lag=7
)


print("\n")
print("=" * 70)

print(
    "SOLAR FLARE ENERGY → "
    "GLOBAL TEMPERATURE ANOMALY"
)

print("=" * 70)

display(
    temp_energy_results
)
```

```python
temp_x_results = correlation_by_lag(
    solar_temp,
    "x_count",
    "temp_anomaly",
    max_lag=7
)


print("\n")
print("=" * 70)

print(
    "X-CLASS FLARES → "
    "GLOBAL TEMPERATURE ANOMALY"
)

print("=" * 70)

display(
    temp_x_results
)
```

```python
temp_m_results = correlation_by_lag(
    solar_temp,
    "m_count",
    "temp_anomaly",
    max_lag=7
)


print("\n")
print("=" * 70)

print(
    "M-CLASS FLARES → "
    "GLOBAL TEMPERATURE ANOMALY"
)

print("=" * 70)

display(
    temp_m_results
)
```

```python
print("\n")
print("=" * 70)
print("TEMPERATURE ANALYSIS SUMMARY")
print("=" * 70)


test_sets = {
    "Flare count":
        temp_flare_results,

    "Flare energy":
        temp_energy_results,

    "X-class flares":
        temp_x_results,

    "M-class flares":
        temp_m_results
}


summary_rows = []


for name, result in test_sets.items():

    if len(result) == 0:
        continue

    best = result.loc[
        result[
            "pearson_r"
        ]
        .abs()
        .idxmax()
    ]

    summary_rows.append({
        "solar_measure":
            name,

        "best_lag_days":
            int(
                best["lag_days"]
            ),

        "pearson_r":
            best["pearson_r"],

        "raw_p":
            best["pearson_p"],

        "fdr_p":
            best["pearson_fdr"]
    })


temperature_summary = pd.DataFrame(
    summary_rows
)


display(
    temperature_summary
)
```

```python
if len(temp_flare_results) > 0:

    best = temp_flare_results.loc[
        temp_flare_results[
            "pearson_r"
        ]
        .abs()
        .idxmax()
    ]


    print(
        "\nStrongest flare-count → "
        "temperature-anomaly relationship:"
    )


    print(
        f"\nLag: "
        f"{int(best['lag_days'])} days"
    )


    print(
        f"Pearson r: "
        f"{best['pearson_r']:.4f}"
    )


    print(
        f"Raw p-value: "
        f"{best['pearson_p']:.6f}"
    )


    print(
        f"FDR-adjusted p-value: "
        f"{best['pearson_fdr']:.6f}"
    )


    print(
        f"Number of observations: "
        f"{int(best['n'])}"
    )


    if best["pearson_fdr"] < 0.05:

        print(
            "\nRESULT: Statistically significant "
            "after FDR correction."
        )

    else:

        print(
            "\nRESULT: NOT statistically significant "
            "after FDR correction."
        )
```
